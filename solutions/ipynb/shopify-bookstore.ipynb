## 1. Shopify Setup and API Credentials

First, set up your Shopify development store and get API credentials.


```javascript
/*
SHOPIFY SETUP STEPS:

1. Create Shopify Partner Account:
   - Go to https://partners.shopify.com
   - Sign up for free
   - Create a development store

2. Create Custom App:
   - In your store admin, go to Settings > Apps and sales channels
   - Click "Develop apps"
   - Click "Create an app"
   - Name it "Bookstore API"
   - Configure API scopes:
     * read_products, write_products
     * read_orders, write_orders
     * read_customers, write_customers
     * read_inventory, write_inventory
   - Install the app
   - Get Admin API access token

3. Get Storefront API Token:
   - In the app, enable Storefront API
   - Get Storefront access token

4. Install Dependencies:
   npm install @shopify/shopify-api
   npm install @shopify/storefront-api-client
   npm install axios dotenv
*/

console.log('Follow the setup steps to get your Shopify credentials');
```

## 2. Environment Configuration

Create a `.env` file with your Shopify credentials.


```javascript
// .env file content:
/*
SHOPIFY_STORE_DOMAIN=your-store.myshopify.com
SHOPIFY_ADMIN_API_TOKEN=shpat_xxxxxxxxxxxxx
SHOPIFY_STOREFRONT_TOKEN=xxxxxxxxxxxxx
SHOPIFY_API_VERSION=2024-01
*/

// Load environment variables
require('dotenv').config();

const shopifyConfig = {
  storeDomain: process.env.SHOPIFY_STORE_DOMAIN,
  adminApiToken: process.env.SHOPIFY_ADMIN_API_TOKEN,
  storefrontToken: process.env.SHOPIFY_STOREFRONT_TOKEN,
  apiVersion: process.env.SHOPIFY_API_VERSION || '2024-01'
};

console.log('Shopify configuration loaded');
```
