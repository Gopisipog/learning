## 2. String Interpolation and Property Binding

Use double curly braces `{{ }}` for string interpolation to display component properties in the template.
Property binding with `[property]` binds component data to element properties.


```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-greeting',
  template: `
    <div>
      <h2>Welcome, {{ name }}!</h2>
      <p>Age: {{ age }}</p>
      <p>Status: {{ isActive ? 'Active' : 'Inactive' }}</p>
      <p>Next year you'll be {{ age + 1 }}</p>
      <img [src]="imageUrl" [alt]="name" />
      <button [disabled]="!isActive">Click Me</button>
    </div>
  `
})
export class GreetingComponent {
  name = 'Gopi';
  age = 30;
  isActive = true;
  imageUrl = 'https://via.placeholder.com/150';
}
```

## 3. Input Properties - Passing Data to Components

Use `@Input()` decorator to pass data from parent to child components. This makes components reusable.


```typescript
import { Component, Input } from '@angular/core';

@Component({
  selector: 'app-user-card',
  template: `
    <div style="border: 1px solid #ccc; padding: 10px; margin: 10px;">
      <h3>{{ name }}</h3>
      <p>Email: {{ email }}</p>
      <p>Role: {{ role }}</p>
    </div>
  `
})
export class UserCardComponent {
  @Input() name: string = '';
  @Input() email: string = '';
  @Input() role: string = '';
}

// Parent component using UserCard
@Component({
  selector: 'app-root',
  template: `
    <div>
      <app-user-card 
        name="Alice" 
        email="alice@example.com" 
        role="Developer">
      </app-user-card>
      <app-user-card 
        name="Bob" 
        email="bob@example.com" 
        role="Designer">
      </app-user-card>
    </div>
  `
})
export class AppComponent {}
```

## 4. Output Properties and Event Emitters

Use `@Output()` and `EventEmitter` to send data from child to parent components.


```typescript
import { Component, Input, Output, EventEmitter } from '@angular/core';

@Component({
  selector: 'app-product',
  template: `
    <div>
      <h4>{{ name }}</h4>
      <p>Price: ${{ price }}</p>
      <button (click)="onAddToCart()">Add to Cart</button>
    </div>
  `
})
export class ProductComponent {
  @Input() name: string = '';
  @Input() price: number = 0;
  @Output() addToCart = new EventEmitter<string>();
  
  onAddToCart() {
    this.addToCart.emit(this.name);
  }
}

// Parent component
@Component({
  selector: 'app-product-list',
  template: `
    <div>
      <app-product 
        name="Laptop" 
        [price]="999" 
        (addToCart)="handleAddToCart($event)">
      </app-product>
      <p>Last added: {{ lastAdded }}</p>
    </div>
  `
})
export class ProductListComponent {
  lastAdded = '';
  
  handleAddToCart(productName: string) {
    this.lastAdded = productName;
    console.log('Added to cart:', productName);
  }
}
```

## 5. Event Binding and Two-Way Binding

Event binding with `(event)` handles user interactions. Two-way binding with `[(ngModel)]` combines property and event binding.


```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-counter',
  template: `
    <div>
      <h3>Count: {{ count }}</h3>
      <button (click)="increment()">Increment</button>
      <button (click)="decrement()">Decrement</button>
      <button (click)="reset()">Reset</button>
      
      <hr>
      
      <input [(ngModel)]="name" placeholder="Enter name" />
      <p>Hello, {{ name }}!</p>
    </div>
  `
})
export class CounterComponent {
  count = 0;
  name = '';
  
  increment() {
    this.count++;
  }
  
  decrement() {
    this.count--;
  }
  
  reset() {
    this.count = 0;
  }
}
```

## 6. Structural Directives - *ngIf

`*ngIf` conditionally adds or removes elements from the DOM based on a boolean expression.


```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-user-status',
  template: `
    <div>
      <h3 *ngIf="isLoggedIn">Welcome back!</h3>
      <h3 *ngIf="!isLoggedIn">Please log in</h3>
      
      <!-- Using else -->
      <div *ngIf="isLoggedIn; else loggedOut">
        <p>You have access to premium features</p>
      </div>
      <ng-template #loggedOut>
        <p>Log in to access premium features</p>
      </ng-template>
      
      <p *ngIf="hasNotifications">🔔 You have new notifications</p>
      
      <button (click)="toggleLogin()">{{ isLoggedIn ? 'Logout' : 'Login' }}</button>
      <button (click)="toggleNotifications()">Toggle Notifications</button>
    </div>
  `
})
export class UserStatusComponent {
  isLoggedIn = false;
  hasNotifications = true;
  
  toggleLogin() {
    this.isLoggedIn = !this.isLoggedIn;
  }
  
  toggleNotifications() {
    this.hasNotifications = !this.hasNotifications;
  }
}
```

## 7. Structural Directives - *ngFor

`*ngFor` repeats a template for each item in a collection. Use `trackBy` for performance optimization.


```typescript
import { Component } from '@angular/core';

interface Todo {
  id: number;
  text: string;
  completed: boolean;
}

@Component({
  selector: 'app-todo-list',
  template: `
    <div>
      <h3>Todo List</h3>
      <ul>
        <li *ngFor="let todo of todos; let i = index; trackBy: trackByTodoId"
            [style.text-decoration]="todo.completed ? 'line-through' : 'none'">
          {{ i + 1 }}. {{ todo.text }}
          <button (click)="toggleTodo(todo)">Toggle</button>
        </li>
      </ul>
      
      <h4>Users</h4>
      <div *ngFor="let user of users">
        <strong>{{ user.name }}</strong> - {{ user.email }}
      </div>
    </div>
  `
})
export class TodoListComponent {
  todos: Todo[] = [
    { id: 1, text: 'Learn Angular', completed: true },
    { id: 2, text: 'Build a project', completed: false },
    { id: 3, text: 'Master TypeScript', completed: false }
  ];
  
  users = [
    { name: 'Alice', email: 'alice@example.com' },
    { name: 'Bob', email: 'bob@example.com' },
    { name: 'Charlie', email: 'charlie@example.com' }
  ];
  
  toggleTodo(todo: Todo) {
    todo.completed = !todo.completed;
  }
  
  trackByTodoId(index: number, todo: Todo): number {
    return todo.id;
  }
}
```

## 8. Structural Directives - *ngSwitch

`*ngSwitch` displays one element from a set of alternatives based on a switch condition.


```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-view-switcher',
  template: `
    <div>
      <button (click)="currentView = 'list'">List View</button>
      <button (click)="currentView = 'grid'">Grid View</button>
      <button (click)="currentView = 'table'">Table View</button>
      
      <div [ngSwitch]="currentView">
        <div *ngSwitchCase="'list'">
          <h4>📋 List View</h4>
          <ul>
            <li>Item 1</li>
            <li>Item 2</li>
            <li>Item 3</li>
          </ul>
        </div>
        
        <div *ngSwitchCase="'grid'">
          <h4>🔲 Grid View</h4>
          <div style="display: grid; grid-template-columns: repeat(3, 1fr); gap: 10px;">
            <div>Item 1</div>
            <div>Item 2</div>
            <div>Item 3</div>
          </div>
        </div>
        
        <div *ngSwitchCase="'table'">
          <h4>📊 Table View</h4>
          <table>
            <tr><td>Item 1</td></tr>
            <tr><td>Item 2</td></tr>
            <tr><td>Item 3</td></tr>
          </table>
        </div>
        
        <div *ngSwitchDefault>
          <p>Please select a view</p>
        </div>
      </div>
    </div>
  `
})
export class ViewSwitcherComponent {
  currentView = 'list';
}
```

## 9. Attribute Directives - ngClass and ngStyle

Attribute directives change the appearance or behavior of elements. `ngClass` and `ngStyle` are commonly used.


```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-styling',
  template: `
    <div>
      <!-- ngClass with object -->
      <p [ngClass]="{ 'active': isActive, 'highlight': isHighlighted }">
        This text has dynamic classes
      </p>
      
      <!-- ngClass with array -->
      <p [ngClass]="['bold', 'italic']">
        This text is bold and italic
      </p>
      
      <!-- ngStyle -->
      <p [ngStyle]="{
        'color': textColor,
        'font-size': fontSize + 'px',
        'background-color': bgColor
      }">
        This text has dynamic styles
      </p>
      
      <button (click)="toggleActive()">Toggle Active</button>
      <button (click)="toggleHighlight()">Toggle Highlight</button>
      <button (click)="changeFontSize()">Change Font Size</button>
    </div>
  `,
  styles: [`
    .active { font-weight: bold; }
    .highlight { background-color: yellow; }
    .bold { font-weight: bold; }
    .italic { font-style: italic; }
  `]
})
export class StylingComponent {
  isActive = false;
  isHighlighted = false;
  textColor = 'blue';
  fontSize = 16;
  bgColor = 'lightgray';
  
  toggleActive() {
    this.isActive = !this.isActive;
  }
  
  toggleHighlight() {
    this.isHighlighted = !this.isHighlighted;
  }
  
  changeFontSize() {
    this.fontSize = this.fontSize === 16 ? 24 : 16;
  }
}
```

## 10. Pipes - Transforming Data

Pipes transform data in templates. Angular provides built-in pipes and you can create custom ones.


```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-pipes-demo',
  template: `
    <div>
      <h3>Built-in Pipes</h3>
      
      <!-- Uppercase/Lowercase -->
      <p>{{ name | uppercase }}</p>
      <p>{{ name | lowercase }}</p>
      <p>{{ name | titlecase }}</p>
      
      <!-- Date pipe -->
      <p>{{ today | date }}</p>
      <p>{{ today | date:'short' }}</p>
      <p>{{ today | date:'fullDate' }}</p>
      <p>{{ today | date:'dd/MM/yyyy' }}</p>
      
      <!-- Currency pipe -->
      <p>{{ price | currency }}</p>
      <p>{{ price | currency:'EUR' }}</p>
      <p>{{ price | currency:'INR':'symbol':'1.0-0' }}</p>
      
      <!-- Number pipe -->
      <p>{{ pi | number:'1.2-5' }}</p>
      
      <!-- Percent pipe -->
      <p>{{ percentage | percent }}</p>
      
      <!-- JSON pipe -->
      <pre>{{ user | json }}</pre>
      
      <!-- Slice pipe -->
      <p>{{ items | slice:0:3 }}</p>
    </div>
  `
})
export class PipesDemoComponent {
  name = 'angular framework';
  today = new Date();
  price = 1234.56;
  pi = 3.14159265359;
  percentage = 0.75;
  user = { name: 'Gopi', age: 30, role: 'Developer' };
  items = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry'];
}
```

## 11. Custom Pipes

Create custom pipes using the `@Pipe` decorator and implementing the `PipeTransform` interface.


```typescript
import { Pipe, PipeTransform, Component } from '@angular/core';

// Custom pipe to reverse a string
@Pipe({
  name: 'reverse'
})
export class ReversePipe implements PipeTransform {
  transform(value: string): string {
    return value.split('').reverse().join('');
  }
}

// Custom pipe to filter array
@Pipe({
  name: 'filter'
})
export class FilterPipe implements PipeTransform {
  transform(items: any[], searchText: string): any[] {
    if (!items || !searchText) {
      return items;
    }
    return items.filter(item => 
      item.toLowerCase().includes(searchText.toLowerCase())
    );
  }
}

// Component using custom pipes
@Component({
  selector: 'app-custom-pipes',
  template: `
    <div>
      <p>{{ 'Angular' | reverse }}</p>
      
      <input [(ngModel)]="searchText" placeholder="Search fruits" />
      <ul>
        <li *ngFor="let fruit of fruits | filter:searchText">
          {{ fruit }}
        </li>
      </ul>
    </div>
  `
})
export class CustomPipesComponent {
  searchText = '';
  fruits = ['Apple', 'Banana', 'Cherry', 'Date', 'Elderberry', 'Fig', 'Grape'];
}
```

## 12. Services and Dependency Injection

Services are singleton classes that provide shared functionality. Use `@Injectable` and inject them via constructor.


```typescript
import { Injectable, Component } from '@angular/core';

// Service
@Injectable({
  providedIn: 'root'
})
export class DataService {
  private data: string[] = ['Item 1', 'Item 2', 'Item 3'];
  
  getData(): string[] {
    return this.data;
  }
  
  addData(item: string): void {
    this.data.push(item);
  }
  
  removeData(index: number): void {
    this.data.splice(index, 1);
  }
}

// Logger service
@Injectable({
  providedIn: 'root'
})
export class LoggerService {
  log(message: string): void {
    console.log(`[LOG] ${new Date().toISOString()}: ${message}`);
  }
}

// Component using services
@Component({
  selector: 'app-data-display',
  template: `
    <div>
      <h3>Data from Service</h3>
      <ul>
        <li *ngFor="let item of items; let i = index">
          {{ item }}
          <button (click)="remove(i)">Remove</button>
        </li>
      </ul>
      <input #newItem placeholder="New item" />
      <button (click)="add(newItem.value); newItem.value=''">Add</button>
    </div>
  `
})
export class DataDisplayComponent {
  items: string[] = [];
  
  constructor(
    private dataService: DataService,
    private logger: LoggerService
  ) {
    this.items = this.dataService.getData();
    this.logger.log('DataDisplayComponent initialized');
  }
  
  add(item: string): void {
    if (item) {
      this.dataService.addData(item);
      this.logger.log(`Added: ${item}`);
    }
  }
  
  remove(index: number): void {
    const item = this.items[index];
    this.dataService.removeData(index);
    this.logger.log(`Removed: ${item}`);
  }
}
```

## 13. HTTP Client - Fetching Data

Use `HttpClient` service to make HTTP requests. Import `HttpClientModule` in your app module.


```typescript
import { Injectable, Component, OnInit } from '@angular/core';
import { HttpClient } from '@angular/common/http';
import { Observable } from 'rxjs';

interface User {
  id: number;
  name: string;
  email: string;
}

// HTTP Service
@Injectable({
  providedIn: 'root'
})
export class UserService {
  private apiUrl = 'https://jsonplaceholder.typicode.com/users';
  
  constructor(private http: HttpClient) {}
  
  getUsers(): Observable<User[]> {
    return this.http.get<User[]>(this.apiUrl);
  }
  
  getUser(id: number): Observable<User> {
    return this.http.get<User>(`${this.apiUrl}/${id}`);
  }
  
  createUser(user: User): Observable<User> {
    return this.http.post<User>(this.apiUrl, user);
  }
}

// Component using HTTP
@Component({
  selector: 'app-user-list',
  template: `
    <div>
      <h3>Users from API</h3>
      <div *ngIf="loading">Loading...</div>
      <div *ngIf="error" style="color: red;">Error: {{ error }}</div>
      
      <ul *ngIf="!loading && !error">
        <li *ngFor="let user of users">
          {{ user.name }} - {{ user.email }}
        </li>
      </ul>
    </div>
  `
})
export class UserListComponent implements OnInit {
  users: User[] = [];
  loading = true;
  error: string | null = null;
  
  constructor(private userService: UserService) {}
  
  ngOnInit(): void {
    this.userService.getUsers().subscribe({
      next: (data) => {
        this.users = data;
        this.loading = false;
      },
      error: (err) => {
        this.error = err.message;
        this.loading = false;
      }
    });
  }
}
```

## 14. Template-Driven Forms

```typescript
import { Component } from '@angular/core';

@Component({
  selector: 'app-login-form',
  template: `
    <div>
      <h3>Login Form</h3>
      <form #loginForm="ngForm" (ngSubmit)="onSubmit(loginForm)">
        <div>
          <label>Username:</label>
          <input 
            type="text" 
            name="username" 
            [(ngModel)]="user.username"
            #username="ngModel"
            required
            minlength="3"
          />
          <div *ngIf="username.invalid && username.touched" style="color: red;">
            <small *ngIf="username.errors?.['required']">Username is required</small>
            <small *ngIf="username.errors?.['minlength']">Min 3 characters</small>
          </div>
        </div>
        
        <div>
          <label>Password:</label>
          <input 
            type="password" 
            name="password" 
            [(ngModel)]="user.password"
            #password="ngModel"
            required
            minlength="6"
          />
          <div *ngIf="password.invalid && password.touched" style="color: red;">
            <small *ngIf="password.errors?.['required']">Password is required</small>
            <small *ngIf="password.errors?.['minlength']">Min 6 characters</small>
          </div>
        </div>
        
        <button type="submit" [disabled]="loginForm.invalid">Login</button>
      </form>
      
      <p>Form Valid: {{ loginForm.valid }}</p>
      <pre>{{ user | json }}</pre>
    </div>
  `
})
export class LoginFormComponent {
  user = {
    username: '',
    password: ''
  };
  
  onSubmit(form: any): void {
    if (form.valid) {
      console.log('Form submitted:', this.user);
      alert(`Logging in as ${this.user.username}`);
    }
  }
}
```

## 15. Reactive Forms

Reactive forms provide more control and are defined in the component class. Import `ReactiveFormsModule`.


```typescript
import { Component, OnInit } from '@angular/core';
import { FormBuilder, FormGroup, Validators } from '@angular/forms';

@Component({
  selector: 'app-registration-form',
  template: `
    <div>
      <h3>Registration Form</h3>
      <form [formGroup]="registrationForm" (ngSubmit)="onSubmit()">
        <div>
          <label>Username:</label>
          <input type="text" formControlName="username" />
          <div *ngIf="username?.invalid && username?.touched" style="color: red;">
            <small *ngIf="username?.errors?.['required']">Username is required</small>
            <small *ngIf="username?.errors?.['minlength']">Min 3 characters</small>
          </div>
        </div>
        
        <div>
          <label>Email:</label>
          <input type="email" formControlName="email" />
          <div *ngIf="email?.invalid && email?.touched" style="color: red;">
            <small *ngIf="email?.errors?.['required']">Email is required</small>
            <small *ngIf="email?.errors?.['email']">Invalid email</small>
          </div>
        </div>
        
        <div>
          <label>Password:</label>
          <input type="password" formControlName="password" />
          <div *ngIf="password?.invalid && password?.touched" style="color: red;">
            <small *ngIf="password?.errors?.['required']">Password is required</small>
            <small *ngIf="password?.errors?.['minlength']">Min 6 characters</small>
          </div>
        </div>
        
        <div>
          <label>Role:</label>
          <select formControlName="role">
            <option value="user">User</option>
            <option value="admin">Admin</option>
            <option value="moderator">Moderator</option>
          </select>
        </div>
        
        <div>
          <label>
            <input type="checkbox" formControlName="agreeToTerms" />
            I agree to terms
          </label>
          <div *ngIf="agreeToTerms?.invalid && agreeToTerms?.touched" style="color: red;">
            <small>You must agree to terms</small>
          </div>
        </div>
        
        <button type="submit" [disabled]="registrationForm.invalid">Register</button>
      </form>
      
      <p>Form Valid: {{ registrationForm.valid }}</p>
      <pre>{{ registrationForm.value | json }}</pre>
    </div>
  `
})
export class RegistrationFormComponent implements OnInit {
  registrationForm!: FormGroup;
  
  constructor(private fb: FormBuilder) {}
  
  ngOnInit(): void {
    this.registrationForm = this.fb.group({
      username: ['', [Validators.required, Validators.minLength(3)]],
      email: ['', [Validators.required, Validators.email]],
      password: ['', [Validators.required, Validators.minLength(6)]],
      role: ['user'],
      agreeToTerms: [false, Validators.requiredTrue]
    });
  }
  
  get username() { return this.registrationForm.get('username'); }
  get email() { return this.registrationForm.get('email'); }
  get password() { return this.registrationForm.get('password'); }
  get agreeToTerms() { return this.registrationForm.get('agreeToTerms'); }
  
  onSubmit(): void {
    if (this.registrationForm.valid) {
      console.log('Form submitted:', this.registrationForm.value);
      alert('Registration successful!');
    }
  }
}
```

## 16. Component Lifecycle Hooks

Angular components have lifecycle hooks that allow you to tap into key moments in a component's lifecycle.


```typescript
import { Component, OnInit, OnDestroy, OnChanges, SimpleChanges, Input } from '@angular/core';

@Component({
  selector: 'app-lifecycle-demo',
  template: `
    <div>
      <h3>Lifecycle Demo</h3>
      <p>Count: {{ count }}</p>
      <p>Message: {{ message }}</p>
      <button (click)="increment()">Increment</button>
      <p>Check console for lifecycle logs</p>
    </div>
  `
})
export class LifecycleDemoComponent implements OnInit, OnDestroy, OnChanges {
  @Input() message: string = '';
  count = 0;
  private intervalId: any;
  
  constructor() {
    console.log('1. Constructor called');
  }
  
  ngOnChanges(changes: SimpleChanges): void {
    console.log('2. ngOnChanges called', changes);
  }
  
  ngOnInit(): void {
    console.log('3. ngOnInit called');
    // Initialize component, fetch data, set up subscriptions
    this.intervalId = setInterval(() => {
      this.count++;
    }, 1000);
  }
  
  ngOnDestroy(): void {
    console.log('4. ngOnDestroy called');
    // Cleanup: unsubscribe, clear timers
    if (this.intervalId) {
      clearInterval(this.intervalId);
    }
  }
  
  increment(): void {
    this.count++;
  }
}
```

## 17. RxJS Observables and Operators

Angular uses RxJS for reactive programming. Observables represent streams of data over time.


```typescript
import { Component, OnInit, OnDestroy } from '@angular/core';
import { Observable, Subject, interval, Subscription } from 'rxjs';
import { map, filter, take, debounceTime, distinctUntilChanged } from 'rxjs/operators';

@Component({
  selector: 'app-observables-demo',
  template: `
    <div>
      <h3>Observables Demo</h3>
      
      <p>Timer: {{ timerValue }}</p>
      
      <div>
        <input 
          type="text" 
          (input)="onSearchChange($event)" 
          placeholder="Search (debounced)"
        />
        <p>Search term: {{ searchTerm }}</p>
      </div>
      
      <button (click)="startTimer()">Start Timer</button>
      <button (click)="stopTimer()">Stop Timer</button>
    </div>
  `
})
export class ObservablesDemoComponent implements OnInit, OnDestroy {
  timerValue = 0;
  searchTerm = '';
  private timerSubscription?: Subscription;
  private searchSubject = new Subject<string>();
  
  ngOnInit(): void {
    // Debounced search
    this.searchSubject.pipe(
      debounceTime(500),
      distinctUntilChanged()
    ).subscribe(term => {
      this.searchTerm = term;
      console.log('Searching for:', term);
    });
  }
  
  startTimer(): void {
    // Create observable that emits every second
    this.timerSubscription = interval(1000).pipe(
      map(n => n + 1),
      filter(n => n <= 10),
      take(10)
    ).subscribe({
      next: (value) => this.timerValue = value,
      complete: () => console.log('Timer complete')
    });
  }
  
  stopTimer(): void {
    this.timerSubscription?.unsubscribe();
  }
  
  onSearchChange(event: any): void {
    this.searchSubject.next(event.target.value);
  }
  
  ngOnDestroy(): void {
    this.timerSubscription?.unsubscribe();
    this.searchSubject.complete();
  }
}
```

## 18. Routing and Navigation

Angular Router enables navigation between views. Define routes and use `routerLink` for navigation.


```typescript
import { NgModule, Component } from '@angular/core';
import { RouterModule, Routes, Router, ActivatedRoute } from '@angular/router';

// Components
@Component({
  selector: 'app-home',
  template: `<h2>Home Page</h2>`
})
export class HomeComponent {}

@Component({
  selector: 'app-about',
  template: `<h2>About Page</h2>`
})
export class AboutComponent {}

@Component({
  selector: 'app-user-detail',
  template: `
    <h2>User Detail</h2>
    <p>User ID: {{ userId }}</p>
    <button (click)="goBack()">Go Back</button>
  `
})
export class UserDetailComponent {
  userId: string | null = null;
  
  constructor(
    private route: ActivatedRoute,
    private router: Router
  ) {
    this.userId = this.route.snapshot.paramMap.get('id');
  }
  
  goBack(): void {
    this.router.navigate(['/']);
  }
}

// Routes configuration
const routes: Routes = [
  { path: '', component: HomeComponent },
  { path: 'about', component: AboutComponent },
  { path: 'user/:id', component: UserDetailComponent },
  { path: '**', redirectTo: '' } // Wildcard route
];

// App component with navigation
@Component({
  selector: 'app-root',
  template: `
    <nav>
      <a routerLink="/" routerLinkActive="active" [routerLinkActiveOptions]="{exact: true}">
        Home
      </a>
      <a routerLink="/about" routerLinkActive="active">About</a>
      <a routerLink="/user/123" routerLinkActive="active">User 123</a>
    </nav>
    
    <router-outlet></router-outlet>
  `,
  styles: [`
    nav a {
      padding: 10px;
      margin: 5px;
      text-decoration: none;
      color: blue;
    }
    nav a.active {
      font-weight: bold;
      color: red;
    }
  `]
})
export class AppComponent {}

// Routing module
@NgModule({
  imports: [RouterModule.forRoot(routes)],
  exports: [RouterModule]
})
export class AppRoutingModule {}
```
